# Simulation de la loi exponentielle par inversion
\n$X=-\frac{1}{\lambda}\ln(U)$ avec $U\sim U(0,1)$. 

```python
import numpy as np
import matplotlib.pyplot as plt

lam=2.0
N=10000
U=np.random.rand(N)
X=-np.log(U)/lam

x=np.linspace(0,np.max(X),500)
f=lam*np.exp(-lam*x)

plt.figure(figsize=(8,5))
plt.hist(X,bins=50,density=True,alpha=0.6,label='Histogramme')
plt.plot(x,f,'r',lw=3,label='Densite theorique')
plt.legend()
plt.grid(True)
plt.show()

print('Moyenne theorique =',1/lam)
print('Moyenne empirique =',np.mean(X))
print('Variance theorique =',1/lam**2)
print('Variance empirique =',np.var(X))
```
