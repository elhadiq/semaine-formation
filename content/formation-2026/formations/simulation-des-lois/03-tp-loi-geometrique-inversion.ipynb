# Simulation de la loi géométrique par la méthode d'inversion

Soit U~U(0,1). On utilise :
$$X=\left\lceil \frac{\ln(1-U)}{\ln(1-p)}\right\rceil.$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import geom

p=0.3
N=100000
U=np.random.rand(N)
#np.ceil(Y renvoie le plus petit entier supérieur ou égal à Y.
X=np.ceil(np.log(1-U)/np.log(1-p)).astype(int)

print("Moyenne empirique:",X.mean())
print("Moyenne théorique:",1/p)
```

```python
k=np.arange(1,21)
#geom.pmf(k,p) calcule les probabilités théoriques de la loi géométrique :
p_theo=geom.pmf(k,p)
#trouver les valeurs distinctes contenues dans le tableau X ;
#compter combien de fois chaque valeur apparaît.
vals,cnts=np.unique(X,return_counts=True)
p_emp=np.array([cnts[vals==i][0]/N if np.any(vals==i) else 0 for i in k])

plt.figure(figsize=(8,5))
plt.bar(k,p_emp,label="Empirique",alpha=0.7)
plt.plot(k,p_theo,"ro-",label="Théorique")
plt.legend(); plt.grid(); plt.show()
```
